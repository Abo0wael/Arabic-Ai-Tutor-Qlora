# Arabic AI Tutor dataset exploration
Inspect the sample before training. These examples validate the pipeline, not model quality.

```python
from pathlib import Path
import json
root = Path.cwd() if (Path.cwd() / 'data').exists() else Path.cwd().parent
rows = [json.loads(line) for line in (root / 'data/sample_dataset.jsonl').read_text(encoding='utf-8').splitlines()]
print('Examples:', len(rows))
print(rows[0]['messages'][2]['content'])
```

```python
lengths = [len(row['messages'][2]['content']) for row in rows]
print({'min_characters': min(lengths), 'max_characters': max(lengths), 'mean_characters': sum(lengths)/len(lengths)})
```
